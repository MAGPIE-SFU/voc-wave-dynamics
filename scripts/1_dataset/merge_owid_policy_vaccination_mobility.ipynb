```python
df_initial = pd.read_csv("owid-covid-data.csv")
```

# ---------------------------------------------------------------------------------------------------------------

## Policy responses

```python
df_workplace = pd.read_csv("workplace-closures-covid.csv")
df_workplace
```

```python
df_school = pd.read_csv("school-closures-covid.csv")
df_school
```

```python
df_internal = pd.read_csv("internal-movement-covid.csv")
df_internal
```

```python
df_international = pd.read_csv("international-travel-covid.csv")
df_international
```

```python
df_stay_home = pd.read_csv("stay-at-home-covid.csv")
df_stay_home
```

```python
df_updated = df_initial.copy()

df_updated = df_updated.merge(df_workplace, on=["Entity", "Code", "Day"], how='left')
df_updated = df_updated.merge(df_school, on=["Entity", "Code", "Day"], how='left')
df_updated = df_updated.merge(df_internal, on=["Entity", "Code", "Day"], how='left')
df_updated = df_updated.merge(df_international, on=["Entity", "Code", "Day"], how='left')
df_updated = df_updated.merge(df_stay_home, on=["Entity", "Code", "Day"], how='left')

df_updated
```

# ---------------------------------------------------------------------------------------------------------------

## Tests & positivity

```python
df_contact_tracing = pd.read_csv("covid-contact-tracing.csv")
df_contact_tracing
```

```python
df_updated_ver2 = df_updated.copy()

df_updated_ver2 = df_updated_ver2.merge(df_contact_tracing, on=["Entity", "Code", "Day"], how='left')

df_updated_ver2
```

# ---------------------------------------------------------------------------------------------------------------

## Vaccinations

```python
df_vaccination_per_capita = pd.read_csv("covid-vaccination-doses-per-capita.csv")
df_vaccination_per_capita
```

```python
df_vaccination_policy = pd.read_csv("covid-vaccination-policy.csv")
df_vaccination_policy
```

```python
df_vaccine_manufacturer = pd.read_csv("covid-vaccine-doses-by-manufacturer.csv")
df_vaccine_manufacturer
```

```python
df_updated_ver3 = df_updated_ver2.copy()

df_updated_ver3 = df_updated_ver3.merge(df_vaccination_per_capita, on=["Entity", "Code", "Day"], how='left')
df_updated_ver3 = df_updated_ver3.merge(df_vaccination_policy, on=["Entity", "Code", "Day"], how='left')
df_updated_ver3 = df_updated_ver3.merge(df_vaccine_manufacturer, on=["Entity", "Code", "Day"], how='left')

df_updated_ver3
```
